# Data exploration and hypothesis testing

**The goal:** Data exploration and t-test

### Imports and data loading




```python
import pandas as pd
from scipy import stats
import numpy as np
```

```python
# Load dataset 
df = pd.read_csv('waze_dataset.csv')
df.head()
```

###  Data exploration and encoding variables

```python
# 1. Create `map_dictionary`
map_dictionary = {'iPhone' : 1, 'Android' : 2}

# 2. Create new `device_type` column
df['device_type'] = df['device']

# 3. Map the new column to the dictionary
df['device_type'] = df['device_type'].map(map_dictionary)
df.head()
```

```python
Iphone = df[df['device_type'] == 1]
Android = df[df['device_type'] == 2]
Iphone['drives'].mean(), Android['drives'].mean()
```


### Task 3. Hypothesis testing

* **Null** -- There is no difference between who use android and who use iphone
* **Alternative** -- there is a difference between who use android and who use iphone

5% as the significance level

```python
# Perform the t-test
stats.ttest_ind(Iphone['drives'], Android['drives'], equal_var = False)
```

**Since pvalue is greater than significance level I fail to reject the null values, so there is no difference between who use the two devices**
